```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.svm import SVC
from sklearn.neural_network import MLPClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score, classification_report
from imblearn.over_sampling import SMOTE
```

```python
# Đọc dữ liệu
data = pd.read_csv(r'F:\Working Sync\LAB projects\Fall Stroke\combined_keypoints.csv')  # Thay 'your_file.csv' bằng tên file của bạn

# Loại bỏ các cột không cần thiết
data = data.drop(['scenario', 'frame_idx'], axis=1)

# Kiểm tra giá trị NaN trong nhãn (position_code)
print("Số lượng giá trị NaN trong nhãn (position_code):", data['position_code'].isna().sum())

# Loại bỏ các dòng có nhãn NaN
data = data.dropna(subset=['position_code'])

# Tách đặc trưng và nhãn
X = data.drop('position_code', axis=1)  # Đặc trưng: 32 cột keypoint
y = data['position_code']  # Nhãn: 0 (không ngã) hoặc 1 (ngã)

# Kiểm tra lại nhãn sau khi loại bỏ NaN
print("Số lượng giá trị NaN trong nhãn sau khi loại bỏ:", y.isna().sum())

# Kiểm tra số lượng giá trị NaN trong đặc trưng trước khi xử lý
print("Số lượng giá trị NaN trong X trước khi xử lý:\n", X.isna().sum())

# Xử lý giá trị NaN trong đặc trưng bằng SimpleImputer (điền bằng 0)
imputer = SimpleImputer(strategy='constant', fill_value=0)
X_imputed = imputer.fit_transform(X)

# Kiểm tra lại sau khi xử lý NaN trong đặc trưng
X_imputed_df = pd.DataFrame(X_imputed, columns=X.columns)
print("Số lượng giá trị NaN trong X sau khi xử lý:\n", X_imputed_df.isna().sum())

# Chuẩn hóa dữ liệu
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_imputed)

# Chia tập dữ liệu
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

# Áp dụng SMOTE để cân bằng dữ liệu
smote = SMOTE(random_state=42)
X_train_balanced, y_train_balanced = smote.fit_resample(X_train, y_train)

# Kiểm tra phân bố nhãn sau khi cân bằng
print("Phân bố nhãn sau SMOTE:\n", pd.Series(y_train_balanced).value_counts())

# Dictionary để lưu kết quả
results = {}
```

```python
from sklearn.linear_model import SGDClassifier
from sklearn.metrics import accuracy_score, classification_report

# Khởi tạo SGDClassifier (mô phỏng SVM tuyến tính)
svm = SGDClassifier(loss='hinge', max_iter=1, warm_start=True, random_state=42)

# Số "epoch" giả lập
n_epochs = 100
n_samples = len(X_train_balanced)

# Huấn luyện từng epoch và hiển thị phần trăm tiến trình
print("Huấn luyện SVM...")
for epoch in range(n_epochs):
    svm.partial_fit(X_train_balanced, y_train_balanced, classes=[0, 1])
    # Tính phần trăm hoàn thành
    progress = (epoch + 1) / n_epochs * 100
    print(f"Đã hoàn thành {progress:.1f}%")

# Dự đoán và đánh giá
y_pred_svm = svm.predict(X_test)
accuracy_svm = accuracy_score(y_test, y_pred_svm)
print("SVM Accuracy:", accuracy_svm)
print("SVM Classification Report:\n", classification_report(y_test, y_pred_svm))
```

```python
import tensorflow as tf
from sklearn.metrics import accuracy_score, classification_report

# Kiểm tra xem GPU có sẵn không
print("GPU Available:", tf.config.list_physical_devices('GPU'))

# Callback tùy chỉnh để hiển thị phần trăm tiến trình
class ProgressCallback(tf.keras.callbacks.Callback):
    def on_epoch_end(self, epoch, logs=None):
        progress = (epoch + 1) / self.params['epochs'] * 100
        print(f"Đã hoàn thành {progress:.1f}%")

# Xây dựng mô hình MLP bằng TensorFlow
model = tf.keras.Sequential([
    tf.keras.layers.Dense(100, activation='relu', input_shape=(X_train_balanced.shape[1],)),
    tf.keras.layers.Dense(50, activation='relu'),
    tf.keras.layers.Dense(1, activation='sigmoid')
])

# Biên dịch mô hình
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# Huấn luyện mô hình trong 100 epoch
print("Huấn luyện MLP...")
history = model.fit(X_train_balanced, y_train_balanced, 
                    epochs=100, 
                    batch_size=32, 
                    validation_data=(X_test, y_test),
                    callbacks=[ProgressCallback()],
                    verbose=0)  # Tắt verbose mặc định để chỉ hiển thị phần trăm

# Dự đoán và đánh giá
y_pred_mlp = (model.predict(X_test) > 0.5).astype(int)
accuracy_mlp = accuracy_score(y_test, y_pred_mlp)
print("MLP Accuracy:", accuracy_mlp)
print("MLP Classification Report:\n", classification_report(y_test, y_pred_mlp))
```

```python
from xgboost import train, DMatrix
from sklearn.metrics import accuracy_score, classification_report
from sklearn.model_selection import train_test_split
from imblearn.over_sampling import SMOTE

# Cân bằng dữ liệu
smote = SMOTE(random_state=42)
X_train_resampled, y_train_resampled = smote.fit_resample(X_train_balanced, y_train_balanced)

# Chia tập validation
X_train, X_val, y_train, y_val = train_test_split(X_train_resampled, y_train_resampled, test_size=0.2, random_state=42)

# Chuẩn bị dữ liệu dạng DMatrix
dtrain = DMatrix(X_train, label=y_train)
dval = DMatrix(X_val, label=y_val)
dtest = DMatrix(X_test)

# Thiết lập tham số
params = {
    'learning_rate': 0.05,
    'max_depth': 5,
    'tree_method': 'hist',
    'device': 'cuda',
    'random_state': 42,
    'scale_pos_weight': 3.45,
    'objective': 'binary:logistic',  # Nếu nhị phân, thay bằng 'multi:softprob' nếu đa lớp
    'eval_metric': 'logloss'         # Metric để đánh giá
}

# Huấn luyện với early stopping
print("Huấn luyện XGBoost...")
bst = train(
    params, 
    dtrain, 
    num_boost_round=1000, 
    evals=[(dval, 'validation')], 
    early_stopping_rounds=10, 
    verbose_eval=10
)

# Dự đoán và đánh giá
y_pred_xgb = bst.predict(dtest)
y_pred_xgb = (y_pred_xgb > 0.5).astype(int)  # Chuyển đổi thành nhãn nếu nhị phân
accuracy_xgb = accuracy_score(y_test, y_pred_xgb)
print("XGBoost Accuracy:", accuracy_xgb)
print("XGBoost Classification Report:\n", classification_report(y_test, y_pred_xgb))
```

```python
from sklearn.ensemble import RandomForestClassifier  # Sửa import
from sklearn.metrics import accuracy_score, classification_report
import numpy as np

# Số cây (tương ứng với số "epoch")
n_estimators = 100
rf_estimators = []

# Huấn luyện từng cây và hiển thị phần trăm tiến trình
print("Huấn luyện Random Forest...")
for i in range(n_estimators):
    # Huấn luyện một cây tại mỗi bước
    rf = RandomForestClassifier(n_estimators=1, max_depth=10, random_state=i)
    rf.fit(X_train_balanced, y_train_balanced)
    rf_estimators.append(rf)
    # Tính phần trăm hoàn thành
    progress = (i + 1) / n_estimators * 100
    print(f"Đã hoàn thành {progress:.1f}%")

# Kết hợp các cây để dự đoán
y_pred_rf = np.zeros(len(X_test), dtype=int)
for rf in rf_estimators:
    y_pred_rf += rf.predict(X_test).astype(int)
y_pred_rf = (y_pred_rf > (n_estimators / 2)).astype(int)  # Bỏ phiếu đa số

# Đánh giá
accuracy_rf = accuracy_score(y_test, y_pred_rf)
print("Random Forest Accuracy:", accuracy_rf)
print("Random Forest Classification Report:\n", classification_report(y_test, y_pred_rf))
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report
import numpy as np

# Khởi tạo mô hình Decision Tree
print("Huấn luyện Decision Tree...")
dt = DecisionTreeClassifier(max_depth=5)

# Huấn luyện toàn bộ dữ liệu một lần
dt.fit(X_train_balanced, y_train_balanced)
print("Đã hoàn thành 100%")

# Dự đoán và đánh giá
y_pred_dt = dt.predict(X_test)
accuracy_dt = accuracy_score(y_test, y_pred_dt)
print("Decision Tree Accuracy:", accuracy_dt)

# Sử dụng zero_division để tránh cảnh báo
print("Decision Tree Classification Report:\n", 
      classification_report(y_test, y_pred_dt, zero_division=0))
```
